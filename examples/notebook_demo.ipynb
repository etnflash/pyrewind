# 노트북 예제
셀들이 순서대로 하나의 프로그램처럼 실행됩니다.

```python
%matplotlib inline
fruits = ["사과", "바나나"]
prices = {"사과": 1500, "바나나": 3000}
```

```python
fruits.append("딸기")
prices["딸기"] = 5000

def total_price(items):
    s = 0
    for it in items:
        s += prices[it]
    return s

print(total_price(fruits))
```
